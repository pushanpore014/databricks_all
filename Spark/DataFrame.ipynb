```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DateType
from datetime import date

schema = StructType([
    StructField("customer_id",  StringType(),  False),
    StructField("name",         StringType(),  True),
    StructField("email",        StringType(),  True),
    StructField("age",          IntegerType(), True),
    StructField("city",         StringType(),  True),
    StructField("signup_date",  DateType(),    True),
])

# Sample Data
data = [
    ("C001", "Alice Sharma",   "alice@example.com",   29, "Mumbai",    date(2024, 1, 15)),
    ("C002", "Rohan Mehta",    "rohan@example.com",   34, "Delhi",     date(2024, 3, 22)),
    ("C003", "Priya Nair",     "priya@example.com",   27, "Bangalore", date(2024, 5, 10)),
    ("C004", "Amit Verma",     "amit@example.com",    41, "Kolkata",   date(2024, 7, 4)),
    ("C005", "Sneha Pillai",   "sneha@example.com",   31, "Chennai",   date(2024, 9, 18)),
]

# Create DataFrame
c_df = spark.createDataFrame(data, schema)

```
